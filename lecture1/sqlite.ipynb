​


DB（SQLite）を使う時の処理の流れ

DBへのコネクションを確立
コネクションオブジェクトからデータ操作用のカーソルオブジェクトを取得
実行したいSQLを定義
SQLを実行
必要があればコミットする
コネクションを閉じる

```python
path = ''
db_name = 'car.db'

# DB接続オブジェクトの作成      
conn = sqlite3.connect(path + db_name)

# SQL(RDBを操作する言語)を実行するためのカーソルオブジェクトの作成
cur = conn.cursor()

# SQL文の作成
# テーブルの作成
sql = 'CREATE TABLE cars (id INT , name TEXT, price REAL)'

#SQL文の実行
cur.execute(sql)

# DB接続の終了
conn.close()
```

### データをDBに挿入する（単一レコード）

```python
path = ''
db_name = 'car.db'

try:
# DB接続オブジェクトの作成      
    conn = sqlite3.connect(path + db_name)

# SQL(RDBを操作する言語)を実行するためのカーソルオブジェクトの取得
    cur = conn.cursor()

# SQL文の作成
# データの挿入
# INSERT INTO テーブル名 (カラム1, カラム2, ...) VALUES (値1, 値2, ...);
# 文字列はシングルクォーテーション(')で囲む
    sql = "INSERT INTO cars (id, name, price) VALUES (1, 'Toyota', 30000);"

#SQL文の実行
    cur.execute(sql)

# 変更をDBに反映
    conn.commit()

except sqlite3.Error as e: 
    print('エラーが発生しました:', e)

finally:    

# DB接続の終了
    conn.close()
```

### データをDBに挿入する（複数行）

```python
path = ''
db_name = 'car.db'

try:
# DB接続オブジェクトの作成      
    conn = sqlite3.connect(path + db_name)

# SQL(RDBを操作する言語)を実行するためのカーソルオブジェクトの取得
    cur = conn.cursor()

# SQL文の作成
# 複数レコードの挿入
# INSERT INTO テーブル名 (カラム1, カラム2, ...) VALUES (列に対応したプレースホルダーをカンマ区切りで並べる);
# 文字列はシングルクォーテーション(')で囲む
    sql = "INSERT INTO cars (id, name, price) VALUES (?, ?, ?);"

    cars =  [
        (2, 'Honda', 28000),
        (3, 'Ford', 32000),
        (4, 'BMW', 55000),
    ]

    
#SQL文の実行
    cur.executemany(sql, cars)

# 変更をDBに反映
    conn.commit()

except sqlite3.Error as e: 
    print('エラーが発生しました:', e)

finally:    

# DB接続の終了
    conn.close()
```

```python
path = ''
db_name = 'car.db'

try:
# DB接続オブジェクトの作成      
    conn = sqlite3.connect(path + db_name)

# SQL(RDBを操作する言語)を実行するためのカーソルオブジェクトの取得
    cur = conn.cursor()

# データを参照するSQL
# SELECT * FROM テーブル名 ;
# * の部分は取得したいカラム名をカンマ区切りで指定することもできる
    sql = "SELECT * FROM cars;"

#SQL文の実行
    cur.execute(sql)

# 変更をDBに反映
# conn.commit()

except sqlite3.Error as e: 
    print('エラーが発生しました:', e)

else:
    for row in cur:
        id, name, price = row
        print(id, name, price)    

finally:    

# DB接続の終了
    conn.close()
```

```python
path = ''
db_name = 'car.db'

try:
# DB接続オブジェクトの作成      
    conn = sqlite3.connect(path + db_name)

# SQL(RDBを操作する言語)を実行するためのカーソルオブジェクトの取得
    cur = conn.cursor()

# データを削除するSQL
# DELETE FROM テーブル名 WHERE id = 1 ;
    sql = "DELETE FROM cars WHERE id = 1; "

#SQL文の実行
    cur.execute(sql)

# 変更をDBに反映
    conn.commit()

except sqlite3.Error as e: 
    print('エラーが発生しました:', e)

finally:    

# DB接続の終了
    conn.close()
```

```python
path = ''
db_name = 'car.db'

try:
# DB接続オブジェクトの作成      
    conn = sqlite3.connect(path + db_name)

# SQL(RDBを操作する言語)を実行するためのカーソルオブジェクトの取得
    cur = conn.cursor()

# データを更新するSQL
# UPDATE テーブル名 SET カラム名 = 更新値 WHERE 条件 ;
    sql = "UPDATE cars SET price = ? WHERE id = ?; "

#SQL文の実行
    cur.execute(sql, (28000, 2))

# 変更をDBに反映
    conn.commit()

except sqlite3.Error as e: 
    print('エラーが発生しました:', e)

finally:    

# DB接続の終了
    conn.close()
```
