# エラーハンドリングの基礎


```python
x = 1
y = 0
```

```python
x / y
# 0で割ってるからエラーになるよね
```

```python
try:    
    result = x / y
except ZeroDivisionError as e :
    result = None
    print("エラーが発生しました。", e)  
    print("0で割ることはできません。結果をNONEに設定します。")

print("結果:", result)
    
```

##　複数のエラーが発生する可能性がある場合


```python
try:
    result = x / z
except ZeroDivisionError as e:
    result = None
    print("エラーが発生しました。:", e)  
    print("0で割ることはできません。結果をNoneに設定します。")    
except NameError as e:
    result = None
    print("エラーが発生しました。:", e)  
    print("変数が定義されていません。結果をNoneに設定します。")

print("結果:", result)
```

```python
z = 3

try:
    result = x / z
except ZeroDivisionError as e:
    result = None
    print("エラーが発生しました。:", e)  
    print("0で割ることはできません。結果をNoneに設定します。")    
except NameError as e:
    result = None
    print("エラーが発生しました。:", e)  
    print("変数が定義されていません。結果をNoneに設定します。")
else:
    print("エラーは発生しませんでした。計算結果を表示します。")

print("結果:", result)
```

## エラーの発生の有無に関わらず実行したい

```python
z = 0

try:
    result = x / z
except ZeroDivisionError as e:
    result = None
    print("エラーが発生しました。:", e)  
    print("0で割ることはできません。結果をNoneに設定します。")    
except NameError as e:
    result = None
    print("エラーが発生しました。:", e)  
    print("変数が定義されていません。結果をNoneに設定します。")
else:
    print("エラーは発生しませんでした。計算結果を表示します。")
finally:                
    print("このブロックは、エラーの発生の有無に関わらず実行されます。")

print("結果:", result)
```

## システム終了以外全てのエラーをハンドリングしたい場合

```python
try:
    result = x / z
except Exception as e:
    result = None
    print("エラーが発生しました。:", e)  
    print("0で割ることはできません。結果をNoneに設定します。")    
else:
    print("エラーは発生しませんでした。計算結果を表示します。")
finally:                
    print("このブロックは、エラーの発生の有無に関わらず実行されます。")

print("結果:", result)
```

## webスクレイピングでのエラー処理

```python
import requests
```

```python
url = 'https://www.musashino-u.ac.jp/invalid-page'  # 存在しないページ
response = requests.get(url)
print("ステータスコード:", response.status_code)

#ステータスコードが200以外の場合に例外を発生させる
response.raise_for_status()
```

```python
url = 'https://www.musashino-u.ac.jp/invalid-page'  # 存在しないページ

try:
    response = requests.get(url)
    response.raise_for_status() 
except requests.exceptions.HTTPError as e:
    print("HTTPエラーが発生しました。:", e)
except requests.exceptions.RequestException as e:
    print("リクエストエラーが発生しました。:", e)
else:
    print("ページの取得に成功しました。")
finally:
    print("処理が完了しました。")
```
